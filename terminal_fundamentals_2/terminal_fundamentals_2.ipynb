````
#@title ##### Terminal Radio: Station Simulator { display-mode: "form" }
%%writefile radio_station/radio.py
#!/usr/bin/env python3

import os
import datetime
import sys
from pathlib import Path
from dataclasses import dataclass


TIME_ENVAR = "RADIO_TIME"
PROGRAM_SCHEDULE_FILE = Path("/content/radio_station/schedules/program_schedule.txt")
REQUESTS_FILE = Path("/content/radio_station/requests/song_requests.csv")
PLAYLIST_DIR = Path("/content/radio_station/playlists")


class IncorrectTimeFormatError(Exception):
  pass


@dataclass
class Song:
  title: str
  artist: str
  album: str


@dataclass
class Program:
  start_time: datetime.time
  end_time: datetime.time
  program_name: str
  playlist: list[Song] | None

  def is_on_air(self, time: datetime.time) -> bool:
    if self.start_time <= self.end_time:
      return self.start_time <= time < self.end_time
    # Overnight show: 23:00-01:00
    return time >= self.start_time or time < self.end_time

  def display(self) -> None:
    print("TERMINAL RADIO")
    print(self.start_time.strftime("%H:%M"))
    print("+++++++++++++++")
    print(f"📻 {self.program_name.title()}")
    print("---------------")
    if self.playlist:
      for song in self.playlist:
        print(f"🎶 {song.title} by {song.artist}")
    else:
      print("No playlist available")
    print("+++++++++++++++")


class Radio:
  program_path = PROGRAM_SCHEDULE_FILE
  playlist_dir = PLAYLIST_DIR

  def __init__(self) -> None:
    self.program_schedule: list[Program] = self._parse_schedule()

  def _parse_schedule(self) -> list[Program]:
    if not self.program_path.exists():
      raise FileNotFoundError(f"Schedule file not found at: {self.program_path}")
    programs = []
    with open(self.program_path, "r") as f:
      for line in f:
        if not line.strip():
          continue
        program_time, program_name = line.split("|")
        start, end = (t.strip() for t in program_time.split("-"))
        program_name = program_name.strip()
        programs.append(Program(
            start_time=datetime.datetime.strptime(start, "%H:%M").time(),
            end_time=datetime.datetime.strptime(end, "%H:%M").time(),
            program_name=program_name,
            playlist=self._parse_playlist(program_name),
        ))
    return programs

  def _parse_playlist(self, program_name: str) -> list[Song] | None:
    filename = program_name.lower().replace(" ", "_")
    playlist_path = self.playlist_dir / f"{filename}.txt"
    if not playlist_path.exists():
      return None
    playlist = []
    with open(playlist_path, "r") as f:
      for line in f:
        if not line.strip():
          continue
        title, artist, album = (part.strip() for part in line.split("|"))
        playlist.append(Song(title=title, artist=artist, album=album))
    return playlist

  def get_current_show(self, time: datetime.time) -> Program | None:
    return next((p for p in self.program_schedule if p.is_on_air(time)), None)


def parse_time_envar() -> datetime.time:
  time = os.environ.get(TIME_ENVAR, None)
  if not time:
    raise ValueError(f"{TIME_ENVAR} environment variable not set")
  try:
    return datetime.datetime.strptime(time, "%H:%M").time()
  except ValueError:
    raise IncorrectTimeFormatError(f"Invalid time format: {time}")


def help_message(radio: Radio) -> None:
  print(f"Usage: {TIME_ENVAR}=HH:MM radio.py")
  print()
  print("Program Schedule:")
  for program in radio.program_schedule:
    start = program.start_time.strftime("%H:%M")
    end = program.end_time.strftime("%H:%M")
    print(f"{start} - {end}: {program.program_name}")


def main() -> None:
  radio = Radio()

  try:
    current_time = parse_time_envar()
  except ValueError:
    help_message(radio)
    sys.exit(0)
  except IncorrectTimeFormatError as e:
    print(e)
    sys.exit(1)

  current_program = radio.get_current_show(current_time)
  if current_program:
    current_program.display()
  else:
    print("No current program")


if __name__ == "__main__":
  main()
````

### **`cut`**

Extract specific parts of each line from a file or text.

````
%%bash
cut -c 5 radio_station/requests/song_requests.csv
````

Extract characters 5 and 10 from each line:


````
%%bash
cut -c 5,10 radio_station/requests/song_requests.csv
````

## **Searching**

There are plenty of tools made for the terminal that allows you to search through files and directories. We'll look at two commonly used commands: `grep` and `find`.


### **`grep`**

Searches for text patterns within files.

> **Styntax**: `grep [options] <pattern> <file_or_directory>`


Let's see if any listeners requested Pop songs by searching the `song_requests.csv` file:


````
%%bash
grep "Pop" radio_station/requests/song_requests.csv
````

Notice that the search is case-sensitive. Use `-i` to ignore case:


````
%%bash
grep -i "pop" radio_station/requests/song_requests.csv
````

You can also exclude lines containing the search term using `-v`:


````
%%bash
grep -iv "pop" radio_station/requests/song_requests.csv
````

#### **Searching Directories Recursively (-r)**

You can also search through all files in a directory and its subdirectories using `-r`.


Let's investigate the radio station logs to find out when shows started:


````
%%bash
grep -ir "start" radio_station/logs
````

### **`find`**

Find files and directories based on their names, types, and other criteria.

> **Styntax**: `find <directory> [options] <expression>`

Let's find the radio station's `readme.txt` file:


````
%%bash
find radio_station -name "readme.txt"
````

`-name` is case-sensative. Use `-iname` for case-insensative search:


````
%%bash
find radio_station -iname "README.TXT"
````

You can also use wildcard patterns to find files matching a particular pattern.


Find all text files:


````
%%bash
find radio_station -name "*.txt"
````

#### **Finding Directories (`-type d`)**

Use `-type d` to search specifically for directories.


Find all directories:


````
%%bash
find radio_station -type d
````

Find directories named "logs":


````
%%bash
find radio_station -type d -name "logs"
````

> 💡 find has many more options for searching files and directories. Check out its manual (`man find`) to explore further.


📒 **Remember the difference**:

- `grep`: What's inside the files?
- `find`: Where are the files?


## **Pipes & Redirection**

Pipes and redirection allow you to control how commands receive input and send output. You can connect commands together, save their output to files, or use files as input.


### **Pipes (`|`)**

Pass the output of one command as input to another command.


Let's start by searching the song requests for Pop songs:


````
%%bash
grep "Pop" radio_station/requests/song_requests.csv
````

Now, let's pipe the output to sort to remove duplicate lines:


````
%%bash
grep "Pop" radio_station/requests/song_requests.csv | sort -u
````

We can keep building on this pipeline. Let's pipe the output to `wc` to count how many unique Pop song requests there are:


````
%%bash
grep "Pop" radio_station/requests/song_requests.csv | sort -u | wc -l
````

### **Output Redirection (`>`, `>>`)**

Send command output to a file instead of displaying it in the terminal.


#### **`>`**

Write output to file, overwriting existing content.


The station manager wants to review the warnings from this morning's broadcasts. Let's save them to a new file:


````
%%bash
grep -i "warning" radio_station/logs/broadcast.log > issues.log
````

````
%%bash
cat issues.log
````

#### **`>>`**

Append output to an existing file without overwriting its contents.


Let's also add the error messages to the same file:


````
%%bash
grep -i "error" radio_station/logs/broadcast.log >> issues.log
````

````
%%bash
cat issues.log
````

> 📒 **Note**: Both `>` and `>>` will create a new file if the file does not yet exist.
